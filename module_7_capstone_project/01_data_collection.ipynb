### Import Required Libraries


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import os
```

### Create Folder Structure

```python
os.makedirs("../data/raw", exist_ok=True)
```

### Define the Web Scraper Function


```python
def scrape_nitter(user, pages=5):
    base_url = "https://nitter.net"
    tweets_data = []

    for page in range(1, pages+1):
        url = f"{base_url}/{user}?page={page}"
        response = requests.get(url, headers={'User-Agent':'Mozilla/5.0'})
        soup = BeautifulSoup(response.text, "html.parser")

        tweets = soup.find_all("div", class_="timeline-item")
        
        for tweet in tweets:
            text = tweet.find("div", class_="tweet-content").text.strip()
            timestamp = tweet.find("span", class_="tweet-date").a['title']

            tweets_data.append({
                "username": user,
                "text": text,
                "timestamp": timestamp
            })

        time.sleep(2) 

    return tweets_data
```

### Scrape Tweets from AI Researchers

```python
users = ["ilyasut", "karpathy", "geoffreyhinton"]
all_data = []

for user in users:
    print(f"Scraping {user}...")
    all_data.extend(scrape_nitter(user, pages=3))
```

### Convert Data to DataFrame and Save

```python
df_raw = pd.DataFrame(all_data)
df_raw=pd.read_csv("data/raw/ai_tweets_raw.csv")
df_raw.head()
```
