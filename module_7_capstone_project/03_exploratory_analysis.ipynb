### Import Libraries and Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("../data/processed/cleaned_tweets.parquet")
```

### View Basic Tweet Statistics

```python
df["user"].value_counts()
```

### Visualize Tweet Frequency per Researcher

```python
df["user"].value_counts().plot(kind="bar")
plt.title("Tweet Count by AI Researcher")
plt.xlabel("Username")
plt.ylabel("Tweet Count")
plt.show()

```
