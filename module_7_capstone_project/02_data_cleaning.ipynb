### Import Libraries

```python
import pandas as pd
import re
import os
```

### Create Folder and Load Raw Data

```python
os.makedirs("../data/processed", exist_ok=True)
df = pd.read_csv("data/raw/ai_tweets_raw.csv")
df.head()
```

### Define Cleaning Function


```python
def clean_text(text):
    text = re.sub(r"http\S+", "", text) 
    text = re.sub(r"[@#]\S+", "", text)
    text = re.sub(r"[^A-Za-z0-9\s]", "", text)
    text = text.lower().strip()
    return text
```


### Apply Cleaning and Save Processed Data

```python
df["clean_text"] = df["text"].apply(clean_text)
df.to_parquet("../data/processed/cleaned_tweets.parquet", index=False)
df.head()
```
