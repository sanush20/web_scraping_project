### Import Libraries and Sample Text

```python
import re

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

import spacy
from textblob import TextBlob
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

nlp = spacy.load("en_core_web_sm")

print("All libraries imported successfully ")
```

```python
texts = [
    "Hello!!! This is my FIRST NLP assignment. I'm super excited!!!",
    "Natural Language Processing (NLP) deals with understanding human language.",
    "Text preprocessing includes cleaning, tokenization, stemming, and lemmatization.",
    "Email me at example123@test.com or visit https://www.example.com for more info.",
]
texts
```

### Build a Text Cleaning Pipeline

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+\.\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

cleaned_texts = [clean_text(t) for t in texts]
cleaned_texts
```

### Tokenization + Stopword Removal (NLTK)

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def tokenize_and_remove_stopwords(text):
    tokens = word_tokenize(text)
    filtered_tokens = [w for w in tokens if w not in stop_words]
    return tokens, filtered_tokens

for i, t in enumerate(cleaned_texts):
    tokens, filtered = tokenize_and_remove_stopwords(t)
    print(f"Original cleaned text {i+1}: {t}")
    print("Tokens:", tokens)
    print("Tokens without stopwords:", filtered)
    print("-" * 50)
```

### Stemming & Lemmatization

```python
ps = PorterStemmer()

def apply_stemming(tokens):
    return [ps.stem(w) for w in tokens]

sample_sentence = "studies studying studied goes going gone"
tokens = word_tokenize(sample_sentence)
stemmed = apply_stemming(tokens)

print("Original tokens:", tokens)
print("Stemmed tokens:", stemmed)
```

### Lemmatization (WordNetLemmatizer)

```python
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

def apply_lemmatization(tokens):
    return [lemmatizer.lemmatize(w) for w in tokens]

lemmatized = apply_lemmatization(tokens)
print("Lemmatized tokens:", lemmatized)
```

### spaCy Linguistic Features (POS & NER)

```python
text_example = texts[1] 

doc = nlp(text_example)

print("Tokens and POS tags:")
for token in doc:
    print(token.text, "->", token.pos_)

print("\nNamed Entities:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

### Feature Extraction(Bag-of-Words & TF-IDF)

```python
print("Cleaned texts:")
for t in cleaned_texts:
    print("-", t)
```

### Bag-of-Words with CountVectorizer

```python
cv = CountVectorizer()
bow_matrix = cv.fit_transform(cleaned_texts)

print("Vocabulary (feature names):")
print(cv.get_feature_names_out())

print("\nBag-of-Words matrix (as array):")
print(bow_matrix.toarray())
```

### TF-IDF with TfidfVectorizer

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(cleaned_texts)

print("TF-IDF feature names:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF matrix:")
print(tfidf_matrix.toarray())
```

### Regex Pattern Recognition

#### Extract emails and URLs from original text

```python
email_pattern = r'\S+@\S+\.\S+'
url_pattern = r'http\S+|www\.\S+'

for t in texts:
    emails = re.findall(email_pattern, t)
    urls = re.findall(url_pattern, t)
    
    print("Original text:", t)
    print("Emails found:", emails)
    print("URLs found:", urls)
    print("-" * 50)
```
